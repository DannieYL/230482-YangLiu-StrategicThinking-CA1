
Skewness Interpretation:
    
1. Skewness < 0: The distribution is negatively skewed (left-skewed).

2. Skewness = 0: The distribution is perfectly symmetrical.

3. Skewness > 0: The distribution is positively skewed (right-skewed).

Degree of Skew:

0 to 0.5: Slightly skewed.
0.5 to 1: Moderately skewed.
1 to 2: Highly skewed.

Greater than 2: Very highly skewed.

```python
df['Entry_tickets'].kurt()
```

Kurtosis Interpretation:


1. Excess kurtosis = 0: The distribution has a kurtosis similar to a normal distribution (mesokurtic).

2. Excess kurtosis > 0: The distribution is leptokurtic, meaning it has a higher, sharper peak and heavier tails than a normal distribution.

3. Excess kurtosis < 0: The distribution is platykurtic, meaning it has a flatter peak and lighter tails compared to a normal distribution.

- Entry_tickets kurtosis is -0.4547 So, excess kurtosis < 0. Hence, we can conclude that the `Entry_tickets` curve is a platykurticcurve. As compared to a normal distribution, its central peak is lower and flatter, and its tails are lighter.

## 3. Inferential Statistics: Hypothesis Testing & Correlation Significance

### Hypothesis Testing

### 1. use Shapiro-Wilk Test for Normality

* ##### Null Hypothesis ($H_0$): `Total_visitors` is normally distributed.
* ##### Alternative Hypothesis ($H_1$): `Total_visitors`  is NOT normally distributed.

```python
import seaborn as sns
import numpy as np
from scipy.stats import shapiro, mannwhitneyu, wilcoxon
```

```python
print(" Shapiro-Wilk Test for 'Total_visitors'")

shap_stat, shap_p = shapiro(df['Total_visitors'])
print(f"Statistic = {shap_stat:.4f}, p-value = {shap_p:.4f}")
if shap_p < 0.05:
    print("→ Reject H₀: Total_visitor is NOT normally distributed\n")
else:
    print("→ Fail to reject H₀: Total_visitor is normally distributed\n")
```

### 2. use Two-tailed, One-sample Z-test for Weekday vs Weekends Visitor footfall

* ##### Null Hypothesis ($H_0$): `Total_visitors` on weekends is the same as the weekdays.
* ##### Alternative Hypothesis ($H_1$): `Total_visitors` on weekends is NOT the same as the weekdays.

#### since the population stand deviation and mean is known, we use Z-test.

```python
# import library packages
import numpy as np
from scipy import stats
from scipy.stats import norm
```

```python
# Step 1: Isolate weekend (1) and weekday (0) visitor counts

weekend_visitors = df.loc[df['Is_Weekend'] == 1, 'Total_visitors'].dropna()
weekday_visitors = df.loc[df['Is_Weekend'] == 0, 'Total_visitors'].dropna()
```

```python
# Step 2 : Compute Z-statistic using weekday mean/SD as reference benchmark

n = len(weekend_visitors)              #sample size

mu0 = weekday_visitors.mean()         # benchmark/population mean
sigma = weekday_visitors.std(ddof=1)  #assumed known benchmark/population SD@@@@@
x_bar = weekend_visitors.mean()      #sample mean

se = sigma / np.sqrt(n)
z_stat = (x_bar - mu0) / se
p_value = 2 * norm.sf(abs(z_stat))
```

```python
# Step 3 : Print results

print(f"Weekday mean (mu0): {mu0:.2f} | Weekday SD (sigma): {sigma:.2f} (n={len(weekday_visitors)})")
print(f"Weekend mean (x_bar): {x_bar:.2f} (n={n})")
print(f"Z-statistic: {z_stat:.4f}")
print(f"Two-tailed p-value: {p_value:.4e}")
```

```python

alpha = 0.05
if p_value < 0.05:
     
    print("→ Reject H₀: `Total_visitors` on weekends is NOT the same as the weekdays.\n")
else:
    print("→ Fail to reject H₀: `Total_visitors` on weekends is the same as the weekdays.\n")
```

### 3. Mann-Whitney U test—and checking correlation p-values for staffing

### Step 1. Weekend vs. Weekday Total_visitors
* #### H0: Mean Total_visitors is equal on weekends and weekdays
* #### H1: Mean Total_visitors differs between weekends and weekdays

```python
from scipy import stats

# Step 1. Weekend vs. Weekday Total_visitors
# H0: Mean Total_visitors is equal on weekends and weekdays
# H1: Mean Total_visitors differs between weekends and weekdays
wknd_vis = df.loc[df['Is_Weekend'] == 1, 'Total_visitors'].dropna()
wkdy_vis = df.loc[df['Is_Weekend'] == 0, 'Total_visitors'].dropna()

t_vis, p_t_vis = stats.ttest_ind(wknd_vis, wkdy_vis, equal_var=False)
u_vis, p_u_vis = stats.mannwhitneyu(wknd_vis, wkdy_vis, alternative='two-sided')

print(f"Weekend visitors (n={len(wknd_vis)}): mean = {wknd_vis.mean():.2f}, median = {wknd_vis.median():.2f}")
print(f"Weekday visitors (n={len(wkdy_vis)}): mean = {wkdy_vis.mean():.2f}, median = {wkdy_vis.median():.2f}")
print(f"Welch t-test: t = {t_vis:.4f}, p = {p_t_vis:.4e}")
print(f"Mann-Whitney U: U = {u_vis:.1f}, p = {p_u_vis:.4e}\n")
```

### Step 2. Weekend vs. Weekday Total_staff

```python

wknd_staff = df.loc[df['Is_Weekend'] == 1, 'Total_staff'].dropna()
wkdy_staff = df.loc[df['Is_Weekend'] == 0, 'Total_staff'].dropna()

t_stf, p_t_stf = stats.ttest_ind(wknd_staff, wkdy_staff, equal_var=False)
u_stf, p_u_stf = stats.mannwhitneyu(wknd_staff, wkdy_staff, alternative='two-sided')

print(f"Weekend staff (n={len(wknd_staff)}): mean = {wknd_staff.mean():.2f}, median = {wknd_staff.median():.2f}")
print(f"Weekday staff (n={len(wkdy_staff)}): mean = {wkdy_staff.mean():.2f}, median = {wkdy_staff.median():.2f}")
print(f"Welch t-test: t = {t_stf:.4f}, p = {p_t_stf:.4e}")
print(f"Mann-Whitney U: U = {u_stf:.1f}, p = {p_u_stf:.4e}\n")
```
